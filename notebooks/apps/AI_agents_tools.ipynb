### designing tools.......

```python
from langchain_core.tools import tool

@tool
def add_num(a:int,b:int):
    """
        it will return the addition
        Args:
        a:num1
        b=num2
    """
    return a+b
# add_num.invoke({'a':10,'b':90})

def mul_num(a:int,b:int):
    """
            it will return the multiplication
            Args:
            a:num1
            b=num2
    """
    return a*b
       
```

### connecting tools with llm 
    (designing agent)
    

```python
from langchain_groq import ChatGroq
from langchain.agents import create_agent
```

```python
llm=ChatGroq(model='qwen/qwen3.6-27b')
```

```python
agent=create_agent(model=llm,tools=[add_num,mul_num],
                   system_prompt="you are a math teacher and use the tools for calculation")
```

###invoking agent

```python
res=agent.invoke({'messages':[{'role':'user','content':'what is 50+70*45'}]})
print(res['messages'][-1].content)
```
